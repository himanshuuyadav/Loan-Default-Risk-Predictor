# 📊 Step 1: Exploratory Data Analysis (EDA)

In this notebook, we will explore the loan dataset to understand distributions, correlations, and key drivers of default risk.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Set style
sns.set(style="whitegrid")
%matplotlib inline
```

## 1. Load Data

```python
DATA_PATH = '../data/raw/accepted_2007_to_2018Q4.csv'

# Loading only a subset for faster exploration
if os.path.exists(DATA_PATH):
    df = pd.read_csv(DATA_PATH, low_memory=False, nrows=100000)
    print(f"Dataset loaded with {df.shape[0]} rows and {df.shape[1]} columns.")
else:
    print("Dataset not found. Please download it using the Kaggle CLI as described in README.md.")
```

## 2. Target Variable Analysis

```python
if 'loan_status' in df.columns:
    plt.figure(figsize=(10, 6))
    sns.countplot(data=df, y='loan_status', order=df['loan_status'].value_counts().index)
    plt.title('Distribution of Loan Status')
    plt.show()
```

## 3. Numeric Feature Distributions

```python
numeric_cols = ['loan_amnt', 'int_rate', 'annual_inc', 'dti', 'fico_range_low']
df[numeric_cols].hist(bins=30, figsize=(15, 10), layout=(2, 3))
plt.tight_layout()
plt.show()
```

## 4. Correlation Analysis

```python
plt.figure(figsize=(10, 8))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Key Numeric Features')
plt.show()
```
